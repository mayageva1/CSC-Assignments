## Pokemon
This dataset contains the name of every pokemon along with 
information about it that could help the user further 
understand its origin and uses. This data was collected 
so that people could better understand the details of
specific pokemon

```python
import pandas as pd
import seaborn as sns
```

```python
pokemon_url = 'https://raw.githubusercontent.com/MainakRepositor/Datasets/master/Pokemon.csv'
pokemon_df = pd.read_csv(pokemon_url,index_col=0)
```

## Question 1: 
Have pokemon overall gotten stronger, weaker, or neither as more generations came out?

```python
sns.set_palette('colorblind')
```

```python
sns.catplot(data=pokemon_df,x='Generation', y='Attack',kind='bar')
```

Interpretation: There does not seem to be a direct correlation between attack power and the generation of pokemon

## Question 2:
Does a pokemon's primary type (type 1) affect its defense?

```python
sns.relplot(data=pokemon_df, x='Type 1', y='Defense',
           hue ='Type 1')
```

Interpretation: For this question, I only looked at the primary type of pokemon, although some of them have two types. It seems as though the defense does somewhat depend on the type. Some types, such as, fariy, fighting, and flying have a very low defense, without a singular pokemon having defense over 100. Some of the types, such electric, steel, and psychic have very high defenses with some occurances of over 200 defense. 

## Question 3:
Do pokemon with an overall higher attack also have a higher special attack?

```python
sns.catplot(data=pokemon_df,x='Attack', y='Sp. Atk')
```

Interpretation: It shows as though there is a very slight to no correlation between the regular attack and special attack
