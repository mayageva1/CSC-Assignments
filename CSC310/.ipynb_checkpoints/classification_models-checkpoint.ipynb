```python
X_train.head()
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
gnb.score(X_test,y_test)
```

```python
pct = 0.1
results = []
for i in range(9):
    gnb = GaussianNB()
    X_train, X_test, y_train, y_test = train_test_split(X,Y,test_size=pct,random_state=5)
    gnb.fit(X_train,y_train)
    train_acc = gnb.score(X_train,y_train)
    test_acc = gnb.score(X_test,y_test)
    results.append({'train_pct': pct,'n_train_samples': len(X_train),'n_test_samples': len(X_test),'train_acc': train_acc,'test_acc': test_acc})
    pct = round(pct + 0.1,1)
```

```python
results_df = pd.DataFrame(results)
results_df
```

Use EDA on this data frame to interpret the results of your experiment. How does training vs test size impact the model’s performance? Does it impact training and test accuracy the same way?

```python
sns.lineplot(data=results_df, x='train_pct', y='train_acc', label='Train Accuracy')
```

```python
sns.lineplot(data=results_df, x='train_pct', y='test_acc', label='Test Accuracy')
```

It seems that the higher percent of training data the model is given, accurate testing will be, which makes sense because with more training data there is more variety and the model learns as it goes, leading to a higher accuracy.
However, the testing accuracy went down significanty at the beginning, and then seemed to stabalize. It seems that the training percentage has an opposite affect on the training accuracy and the testing accuracy.
