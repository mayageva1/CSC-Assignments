# Pokemon
This dataset contains the name of every pokemon along with information about it that could help the user further understand its origin and uses. This data was collected so that people who like these games can have more knowledge about it. 
[More info](https://github.com/MainakRepositor/Datasets/blob/master/Pokemon.csv)

# National Parks
This dataset contains the name of national parks in the US, along with their code, state, size, and coordinates of where to find it. This data was created for peopel who are interested in visiting some or all of the US national parks, and probably to better organize which parks to visit when they are in a specific state.
[More info](https://github.com/MainakRepositor/Datasets/blob/master/parks.csv)

# Github Issues
This dataset is a collection of issues that have occured in Github. It is organized in chronological order and tells us the language the issue was found in, along with the quarter and count of issues. This dataset is probably used for anyone who is interested in which programming languages have the most issues or when certain issues got fixed.
[More info](https://github.com/MainakRepositor/Datasets/blob/master/Github_Stats/issues.csv)
