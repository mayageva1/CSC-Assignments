```python
from sklearn.feature_extraction import text
from sklearn.metrics.pairwise import euclidean_distances
from sklearn import datasets
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.naive_bayes import MultinomialNB, GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import confusion_matrix
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV, cross_val_score
```

Create a dataframe using the given file and change the fake/real label to 0 and 1 to be able to compare them.

```python
path = '/Users/mayageva/Downloads/fake_or_real_news.csv'
df = pd.read_csv(path)
df = df.sample(5000, random_state=1)
df['label'] = df['label'].map({'FAKE': 0, 'REAL': 1})
df.head()
```

## Is the text or the title of an article more predictive of whether it is real or fake? vectorizing the text and fitting it with tfidf

Creating the text and titles vectorizer, then fitting tfidf

```python
text_vectorizer = CountVectorizer()
text_vec = text_vectorizer.fit_transform(df['text'])
tfidf = text.TfidfTransformer()
text_tfidf = tfidf.fit_transform(text_vec)
title_vectorizer = CountVectorizer()
title_vec = title_vectorizer.fit_transform(df['title'])
title_tfidf = tfidf.fit_transform(title_vec)
```

Split and test data for both text and title

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    text_tfidf, df['label'], test_size=0.2, random_state=1
)
X_train_title, X_test_title = train_test_split(
    title_tfidf, test_size=0.2, random_state=1
)
```

comparing Guassian NB and MultinomiaNB in order to predect whether a articles is real/fake.

optimizing multinomialNB for text

```python
param_mnb = {'alpha': [0.1, 0.5, 1.0, 2.0, 5.0]}
grid_search_opt_text = GridSearchCV(MultinomialNB(), param_mnb)
grid_search_opt_text.fit(X_train_text, y_train)
```

fitting the text with multinomiaNB and printing the accuracy score

```python
opt_mnb_text = grid_search_opt_text.best_estimator_
opt_mnb_text.fit(X_train_text, y_train)
y_pred_opt_mnb_text = opt_mnb_text.predict(X_test_text)
accuracy_score(y_test, y_pred_opt_mnb_text)
```

optimizing multinomialNB for title

```python
grid_search_opt_title = GridSearchCV(MultinomialNB(), param_mnb)
grid_search_opt_title.fit(X_train_title, y_train)
```

fitting the title with multinomiaNB and printing the accuracy score

```python
opt_mnb_title = grid_search_opt_title.best_estimator_
opt_mnb_title.fit(X_train_title, y_train)
y_pred_opt_mnb_title = opt_mnb_title.predict(X_test_title)
accuracy_score(y_test, y_pred_opt_mnb_title)
```

Train GuassianNB on text and print accuracy score

```python
#I got an error when trying to fit gaussianNB and it said to use .toarray()
X_train_text_dense = X_train_text.toarray()
X_test_text_dense = X_test_text.toarray()
X_train_title_dense = X_train_title.toarray()
X_test_title_dense = X_test_title.toarray()
```

Train GuassianNB on title and print accuracy score

```python
gnb_text = GaussianNB()
gnb_text.fit(X_train_text_dense, y_train)
y_pred_text_gnb = gnb_text.predict(X_test_text_dense)
accuracy_score(y_test, y_pred_text_gnb)
```

Train GuassianNB on title and print accuracy scire

```python
gnb_title = GaussianNB()
gnb_title.fit(X_train_title_dense, y_train)
y_pred_title_gnb = gnb_title.predict(X_test_title_dense)
accuracy_score(y_test, y_pred_title_gnb)
```

As we can see, MultinomiaNB is the stronger choice for this dataset. It has accuracy scores of 0.905 for text and 0.802 for titles, while GaussianNB has 0.804 for text and 0.694 for titles. Although they have the same score for text, MultinomialNB has a much stronger accuracy for the titles.

With MultiomialNB, it was clear that the model was better able to accuracely predict whether an article is fake or real based on its title, rather than text. In the real world application of this dataset, this means that the titles of fake/real articles follow a more similar structure/words in their titles than the actual content(the text). 

There is a clear answer that the title of an article is much more predictive of whether its fake or real than the text.

I dont believe that this analysis is currently strong enough to for sure claim that the titles of articles are more telling of whether theyre fake/real than their text. In order to be more sure, I would repeat this entire process of fitting and comparing models using a different classifier, such as svm, and compare the results to see if they gave the same outcome. I also would use cross validation to have a clearer answer.

## Are titles of real or fake news more similar to one another?

```python
df = df.reset_index(drop=True)
real_titles = df[df['label'] == 1].index
fake_titles = df[df['label'] == 0].index
subset_real = real_titles[:50] 
subset_fake = fake_titles[:50]
subset_rows = np.concatenate([subset_real, subset_fake])
sns.heatmap(euclidean_distances(title_tfidf[subset_rows]))
```

Calculating the avergae euclidean distance of real titles

```python
real_titles_tfidf = title_tfidf[real_titles]
euclidean_dist_real = euclidean_distances(real_titles_tfidf)
np.mean(euclidean_dist_real)
```

Calculating the avergae euclidean distance of fake titles

```python
fake_titles_tfidf = title_tfidf[fake_titles]
euclidean_dist_fake = euclidean_distances(fake_titles_tfidf)
np.mean(euclidean_dist_fake)
```

This analysis contains a randomly selected sample of 50 of each fake and real articles.

The first 50 items (top-right quarter of the heatmap) have a slightly darker colored heatmap coloring. This indicates that the Euclidean distance between the titles is smaller, meaning that there is a more similarity between the titles. This section represents the euclidean distance between the subset of real and fake titles, which can tell us that these titles have more similar wording to one another.

Furthermore, I calculated the mean euclidean distance for the real and fake titles. For the real titles, it was about 1.4044, and about 1.4054 for the fake. These numbers are extremely close, which means that there isnt too much of a difference in the wording between fake titles to eachother and real titles to eachother.

Overall, There is a slightly higher similarity between real news titles than fakes ones.

I believe that this randomly selected sample of 50 of each fake and real article is not the strongest proof of the accuracy of similarity between real/fake titles.
