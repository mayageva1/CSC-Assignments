Computes a grade for CSC/DSP310 from numbers of achievements at each level

```python
def compute_grade(num_level1, num_level2, num_level3):
    if num_level3 >= 15 and num_level2 >= 15 and num_level1 >= 15:
        return 'A'
    elif num_level3 >= 10 and num_level2 >= 15 and num_level1 >= 15:
        return 'A-'
    elif num_level3 >= 5 and num_level2 >= 15 and num_level1 >= 15:
        return 'B+'
    elif num_level3 >= 0 and num_level2 >= 15 and num_level1 >= 15:
        return 'B'
    elif num_level3 >= 0 and num_level2 >= 10 and num_level1 >= 15:
        return 'B-'
    elif num_level3 >= 0 and num_level2 >= 5 and num_level1 >= 15:
        return 'C+'
    elif num_level3 >= 0 and num_level2 >= 0 and num_level1 >= 15:
        return 'C'
    elif num_level3 >= 0 and num_level2 >= 0 and num_level1 >= 10:
        return 'C-'
    elif num_level3 >= 0 and num_level2 >= 0 and num_level1 >= 5:
        return 'C'
    else: return 'D'
```

```python
compute_grade(15,15,15)
```

```python
compute_grade(15,15,13)
```

```python
compute_grade(15,14,14)
```

```python
compute_grade(14,14,14)
```

```python
compute_grade(4,3,1)
```

```python
compute_grade(15,15,6)
```

```python
assert compute_grade(15,15,15) == 'A'

assert compute_grade(15,15,13) == 'A-'

assert compute_grade(15,14,14) == 'B-'

assert compute_grade(14,14,14) == 'C-'

assert compute_grade(4,3,1) == 'D'

assert compute_grade(15,15,6) =='B+'
```
