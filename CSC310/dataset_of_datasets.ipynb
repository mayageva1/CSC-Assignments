```python
import pandas as pd
```

```python
from datasets import dataset_list
```

iterate through each csv file and find its name, source, url, number of rows, number of columns, and number of numerical values in each of them. It then outputs this information to a dataframe

```python
metadata = []
for dataset_dict in dataset_list:
    current_set = dataset_dict['load_func'](dataset_dict['url'])
    current_set.to_csv(dataset_dict['name'])
    
    columns = list(current_set.columns)
    numerical_counter = 0
    
    for c in columns:
        if current_set[c].dtype == 'int64' or current_set[c].dtype == 'float64':
            numerical_counter += 1
                          
    metadata.append({'name':dataset_dict['name'],'source':dataset_dict['url'],
    'num_rows':len(current_set),'num_columns':len(current_set.columns),
    'num_numerical':numerical_counter})
    
df = pd.DataFrame(data=metadata, index=None, columns=None, dtype=None, copy=None)
```
